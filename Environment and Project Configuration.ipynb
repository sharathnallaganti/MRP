```python
# Create folders for documents, vectors, evidence, models, and reports.


from pathlib import Path

PROJECT_ROOT = Path("/kaggle/working/taxify")

FOLDERS = {
    "documents": PROJECT_ROOT / "documents",
    "processed": PROJECT_ROOT / "processed",
    "vectors": PROJECT_ROOT / "vectors",
    "evidence": PROJECT_ROOT / "evidence",
    "reports": PROJECT_ROOT / "reports",
    "models": PROJECT_ROOT / "models",
}

for folder in FOLDERS.values():
    folder.mkdir(parents=True, exist_ok=True)

for name, path in FOLDERS.items():
    print(f"{name}: {path} | exists: {path.exists()}")
```

```python
# Save project configuration

# Save one reproducible configuration for all Taxify notebooks.
import json
import random
import numpy as np

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

CONFIG = {
    "project_name": "Taxify",
    "project_title": (
        "Evidence-Driven Security Evaluation for LLM, "
        "RAG, and Tool-Using Applications"
    ),
    "case_study": "United States federal taxation",
    "llm_model": "meta-llama/Llama-3.2-3B-Instruct",
    "embedding_model": "sentence-transformers/all-MiniLM-L6-v2",
    "vector_database": "FAISS",
    "chunk_size_words": 180,
    "chunk_overlap_words": 35,
    "retrieval_top_k": 4,
    "security_modes": ["vulnerable", "secured"],
    "roles": ["taxpayer", "preparer", "administrator"],
    "vulnerabilities": [
        "direct_prompt_injection",
        "indirect_prompt_injection",
        "sensitive_information_disclosure",
        "rag_knowledge_base_poisoning",
        "unauthorized_retrieval",
        "insecure_output_handling",
        "excessive_agency"
    ],
    "seed": SEED,
    "real_taxpayer_data_allowed": False,
    "budget_usd": 0
}

config_path = PROJECT_ROOT / "taxify_config.json"

with config_path.open("w", encoding="utf-8") as file:
    json.dump(CONFIG, file, indent=2)

print(json.dumps(CONFIG, indent=2))
print("\nConfiguration saved:", config_path)
print("File exists:", config_path.exists())
```
